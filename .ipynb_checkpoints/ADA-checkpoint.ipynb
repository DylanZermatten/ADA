```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    mean_squared_error, mean_absolute_error,
    accuracy_score, confusion_matrix,
    roc_auc_score, roc_curve
)


# GENERAL PARAMETERS

START_DATE = "2017-11-20"
END_DATE   = "2025-01-01"

WINDOW_VOL  = 30   # window for realized volatility (rolling std)
WINDOW_CORR = 30   # window for rolling correlation
HORIZON     = 5    # forecasting horizon (e.g. predict 5 days ahead)



# Download raw price data

# Yahoo Finance tickers: SP500, Bitcoin, Gold, VIX
tickers = ["^GSPC", "BTC-USD", "GC=F", "^VIX"]

raw = yf.download(
    tickers,
    start=START_DATE,
    end=END_DATE,
    auto_adjust=True,
    progress=False
)["Close"]

# Rename columns to simple asset names
raw.columns = ["SP500", "BTC", "Gold", "VIX"]


# Align data and drop incomplete days


# Force business-day frequency (Mon–Fri)
df = raw.asfreq("B")

# Keep only days where all 4 assets have real values
cols_assets = ["SP500", "BTC", "Gold", "VIX"]
df = df.dropna(subset=cols_assets)


# A3 — returns, volatility, correlation

# Log prices
log_prices = np.log(df[cols_assets])

# Log returns
for col in cols_assets:
    df[f"ret_{col}"] = log_prices[col].diff()

# volatility
for col in cols_assets:
    df[f"vol_{col}"] = df[f"ret_{col}"].rolling(WINDOW_VOL).std()

# correlation BTC–SP500
df["corr_BTC_SP500"] = (
    df["ret_BTC"].rolling(WINDOW_CORR).corr(df["ret_SP500"])
)

# Annualized volatility (optional, based on 252 trading days)
for col in cols_assets:
    df[f"vol_ann_{col}"] = df[f"vol_{col}"] * np.sqrt(252)

# Drop initial NaNs from diff/rolling windows
df_base = df.dropna()


# Basic visualizations

# Normalized prices (base 100)
assets = ["SP500", "BTC", "Gold"]
norm_prices = df_base[assets].divide(df_base[assets].iloc[0]) * 100

plt.figure(figsize=(12, 6))
for col in assets:
    plt.plot(norm_prices.index, norm_prices[col], label=col)

plt.title("Normalized Price Evolution (Base 100)")
plt.xlabel("Date")
plt.ylabel("Index (Base 100)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# VIX level
plt.figure(figsize=(12, 4))
plt.plot(df_base.index, df_base["VIX"])
plt.title("VIX Index Level")
plt.xlabel("Date")
plt.ylabel("VIX")
plt.grid(True)
plt.tight_layout()
plt.show()
```
